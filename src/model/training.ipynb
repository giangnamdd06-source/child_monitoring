```python

from ultralytics import YOLO


def main():
    model = YOLO("yolo11n.pt")

    model.train(
        data="data.yaml",   
        epochs=100,        
        imgsz=640,          
        batch=16,          
        patience=50,        
        workers=4,
        device=0,           
        project="runs",
        name="yolo11_advanced",
    )

    metrics = model.val()
    print("mAP50    :", metrics.box.map50)
    print("mAP50-95 :", metrics.box.map)
    results = model.predict(r"C:\Users\LENOVO\do_an\dataset_gray\train_and_validation\test\images", conf=0.5, save=True)
    for r in results:
        for box in r.boxes:
            print("lop:", int(box.cls), "tin cay:", float(box.conf), "khung:", box.xyxy.tolist())


if __name__ == "__main__":
    main()
```
